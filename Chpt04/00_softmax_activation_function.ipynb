```python
import numpy as np
import math

layer_outputs = [4.8, 1.21, 2.385]

E = math.e

exp_values = []
for output in layer_outputs:
    exp_values.append(E ** output)

exp_values
```

```python
# Now we need to normalize 

norm_sum = sum(exp_values)

norm_values = []
for value in exp_values:
    norm_values.append(value / norm_sum)


norm_values
```
