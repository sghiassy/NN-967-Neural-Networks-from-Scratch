```python
import numpy as np

# Starting Numbers
#x0 ─┬──────────────┐
#    │              │
#    │              ▼
#    │        ┌──────────┐
#    │        │ Hidden   │
#    │        │ Neuron 0 │──┐
#    │        └──────────┘  │
#    │                      │
#    │                      ▼
#    │                    ┌───────┐
#    │                    │       │
#    │                    │ Output│──► y ─► L
#    │                    │Neuron │
#    │                    │       │
#    │                    └───────┘
#    │                      ▲
#    │                      │
#    │        ┌──────────┐  │
#    │        │ Hidden   │──┘
#    │        │ Neuron 1 │
#    │        └──────────┘
#    │
#x1 ─┴──────────────────────


x0 = 1.2
x1 = -0.7
x = np.array([[x0, x1]])
w00 = 0.5
w10 = 1.0
w1 = np.array([[w00, w10]])
w11 = 0.9
w01 = -0.4
w2 = np.array([[w01, w11]])
w_1 = np.array([[0.5, -0.4], [1.0, 0.9]], dtype=float)
b10 = 0.2
b11 = 1.3
b1 = np.array([[b10, b11]])
v0 = 1.4
v1 = -0.8
w_2 = np.array([[v0, v1]]).T
b2 = 0.05
y_true = 0.7
w_1
```

```python
z1 = x0 * w00 + x1 * w10 + b10
z1
print(z1)
z1 == np.dot(x, w1.T) + b10
```

```python
z2 = x0 * w01 + x1 * w11 + b11
z2
print(z2)
z2 == np.dot(x, w2.T) + b11
```

```python
z = np.dot(x, w_1) + b1
z
```

```python
h0 = z1 if z1 > 0 else 0
h0
```

```python
h1 = z2 if z2 > 0 else 0
h1
```

```python
h = np.array([[h0, h1]])
h
```

```python
y = h0 * v0 + h1 * v1 + b2
y
print(y)
y == np.dot(z, w_2) + b2
```

```python
L = (1/2)*(y - y_true) ** 2
```
